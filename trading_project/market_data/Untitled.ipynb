```python
import json
import websocket
import time
import threading

# WebSocket URL and authentication details
WS_URL = "wss://smartapisocket.angelone.in/smart-stream"
API_KEY = "your_api_key"
CLIENT_CODE = "your_client_code"
FEED_TOKEN = "your_feed_token"
JWT_TOKEN = "your_jwt_token"

# Subscription request data
SUBSCRIBE_REQUEST = {
    "correlationID": "abcde12345",
    "action": 1,  # 1 for Subscribe
    "params": {
        "mode": 1,  # 1 for LTP
        "tokenList": [
            {"exchangeType": 1, "tokens": ["10626", "5290"]},  # Example tokens for NSE_CM
        ]
    }
}

# Heartbeat message
HEARTBEAT_MESSAGE = "ping"

# Request headers for WebSocket connection
def on_open(ws):
    print("WebSocket connection established.")
    
    # Send authentication details
    auth_request = {
        "Authorization": f"Bearer {JWT_TOKEN}",
        "x-api-key": API_KEY,
        "x-client-code": CLIENT_CODE,
        "x-feed-token": FEED_TOKEN
    }
    
    ws.send(json.dumps(SUBSCRIBE_REQUEST))
    print("Subscribed to tokens:", SUBSCRIBE_REQUEST["params"]["tokenList"])

# Send heartbeat message every 30 seconds to keep the connection alive
def send_heartbeat(ws):
    while True:
        time.sleep(30)
        ws.send(HEARTBEAT_MESSAGE)
        print("Heartbeat sent.")

# Function to handle incoming WebSocket messages
def on_message(ws, message):
    print("Received message:", message)

# Function to handle errors
def on_error(ws, error):
    print("Error:", error)

# Function to handle WebSocket close
def on_close(ws, close_status_code, close_msg):
    print("WebSocket closed with status:", close_status_code)

# WebSocket connection setup
def start_websocket():
    # WebSocket App with the given URL
    ws = websocket.WebSocketApp(
        WS_URL,
        on_open=on_open,
        on_message=on_message,
        on_error=on_error,
        on_close=on_close
    )
    
    # Start heartbeat thread
    heartbeat_thread = threading.Thread(target=send_heartbeat, args=(ws,))
    heartbeat_thread.daemon = True
    heartbeat_thread.start()
    
    # Run the WebSocket
    ws.run_forever()

# Start the WebSocket connection
if __name__ == "__main__":
    start_websocket()
```
